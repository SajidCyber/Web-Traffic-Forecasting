```python
# =================================================================
# NOTEBOOK: 01_data_exploration.ipynb
# GOAL: Load, clean, analyze, and save the initial dataset.
# =================================================================

import pandas as pd
import matplotlib.pyplot as plt
import os

# --- 1. File Paths and Setup ---
# Set the project base path
BASE_PATH = 'C:\\Users\\acer\\OneDrive\\Desktop\\AimlWebforecasting'
DATA_RAW_PATH = f'{BASE_PATH}\\data\\raw\\smght.csv' 
DATA_PROCESSED_PATH = f'{BASE_PATH}\\data\\processed\\cleaned_data.pkl'
PLOT_DIR = f'{BASE_PATH}\\reports\\figures'

# Create necessary output directories
os.makedirs(f'{BASE_PATH}\\data\\processed', exist_ok=True)
os.makedirs(PLOT_DIR, exist_ok=True)

# Load the raw dataset
try:
    df = pd.read_csv(DATA_RAW_PATH)
    print("--- Data loaded successfully ---")
except FileNotFoundError:
    print(f"Error: File not found at {DATA_RAW_PATH}. Check 'data/raw/'.")
    exit()

# --- 2. Data Inspection and Cleaning ---

print("\nDataset Shape:", df.shape)
df.info()

# Convert 'date' to datetime, set as index, and sort
df['date'] = pd.to_datetime(df['date'])
df.set_index('date', inplace=True)
df.sort_index(inplace=True)

# Rename the target column
df.rename(columns={'web_traffic': 'y'}, inplace=True)
df.drop_duplicates(inplace=True)

print("\n--- Cleaned Data Snapshot ---")
print(df.head())

# --- 3. Data Quality Check ---

print("\n--- Missing Value Report ---")
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

if missing_values.empty:
    print("✅ No missing values found.")
else:
    print("⚠️ Missing values detected. To be handled in the next step:\n", missing_values)

# --- 4. Exploration: Target Visualization ---

# Plot the time series of web traffic
plt.figure(figsize=(14, 6))
df['y'].plot(title='Web Traffic (y) Over Time', color='darkblue', linewidth=1.5)
plt.xlabel("Date")
plt.ylabel("Web Traffic (Customers)")
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# Save the plot
plot_path = f'{PLOT_DIR}\\01_target_timeseries.png'
plt.savefig(plot_path)
plt.close() 
print(f"\nTime-series plot saved to '{plot_path}'.")

# --- 5. Initial Correlation Assessment ---

print("\n--- Top 5 Features Correlated with Target (y) ---")
# Check correlations for initial feature assessment
correlation_matrix = df.corr(numeric_only=True)['y'].sort_values(ascending=False)
print(correlation_matrix.head(6))

# --- 6. Output Cleaned Data ---

# Save the DataFrame for use in the next notebook (02_feature_engineering)
df.to_pickle(DATA_PROCESSED_PATH)
print(f"\nCleaned data saved to '{DATA_PROCESSED_PATH}'.")

# =================================================================
```
